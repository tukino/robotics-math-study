## Where is the forward kinematics? / 順運動学はどこか

`forward_kinematics_1link()` is the forward-kinematics calculation. It receives the known link length and joint angle. It converts them into the endpoint `(x, y)` using `L*cos(θ)` and `L*sin(θ)`. The drawing code only visualizes that result.  
`forward_kinematics_1link()`が順運動学の計算です。既知のリンク長と関節角を受け取ります。それらを`L*cos(θ)`と`L*sin(θ)`で先端座標`(x, y)`へ変換します。描画コードは、その計算結果を可視化しているだけです。

## Numerical check / 数値確認

The matrix multiplies the same local points at each angle. Compare these values with the red and green points in the figure.  
各角度で、同じローカル点に回転行列を掛けます。結果を図中の赤と緑の点と比較します。

```python
local_x = np.array([1.0, 0.0])
local_y = np.array([0.0, 1.0])

for theta_deg in (0, 30, 45, 90):
    rotation = rotation_matrix_2d(theta_deg)
    x_axis_world = rotation @ local_x
    y_axis_world = rotation @ local_y
    print(f"θ={theta_deg:>3}°  local (1,0) -> {x_axis_world}  local (0,1) -> {y_axis_world}")
```

## Why are these the columns of the matrix? / なぜ回転行列の列になるのか

A matrix-vector product is a weighted sum of the matrix columns. Multiplying by `(1, 0)` means `1 × first column + 0 × second column`, so the result is the first column. Multiplying by `(0, 1)` selects the second column in the same way. The two columns therefore show where the local x and y basis vectors moved in the world frame.  
行列とベクトルの積は、行列の各列の重み付き和です。`(1, 0)`を掛けると`第1列×1 + 第2列×0`となり、第1列がそのまま残ります。`(0, 1)`は同様に第2列を選びます。したがって2つの列は、ローカルx、y基底が世界座標系でどこへ移ったかを表します。
